```python
#1. numpy
# pip install numpy

#2.  일반적인 파이썬의 리스트와 넘파이 배열의 차이
aaa= [10,20,30]
bbb= [4,5,6]
print(aaa)
print(bbb)

#리스트의 + 연산.. (덧셈이 아니라 연결연산자)
ccc= aaa + bbb
print(ccc)
```

```python
# 넘파이 배열
import numpy as np

aa= np.array([10,20,30])
bb= np.array([4,5,6])

# 배열끼리 + 연산 -- 벡터값의 산술연산이 됨. -- 같은 자리의 요소들끼리 덧셈
cc= aa + bb
print(cc)
```

```python
# 다차원 배열일때..
aaaa= np.array([ [10,20,30,40],[5,6,7,8] ])
print(aaaa.shape) #(행,열)

bbbb= np.array([
    [100,200,300,400],
    [50,60,70,80]
])
print(bbbb.shape)

cccc= aaaa+bbbb
print(cccc)
```

```python
#곱셈도 됨 (합성 곱 .. 요소끼리 곱)
print( aaaa - bbbb )
print( aaaa * bbbb ) # 행렬끼리 곱... 합성곱.. CNN
print( aaaa / bbbb )
```

```python
aaa=[10,20,30]
print(aaa*2)

bbb= np.array([10,20,30])
print(bbb*2) #벡터와 스칼라의 곱 -- 요소마다 2씩 곱해짐.
```

```python
aaa= [10,20,30]
print(aaa)

#리스트 --> 넘파이 배열
bbb= np.array(aaa)
print(bbb)

print(type(aaa))
print(type(bbb))

#요소들의 자료형을 확인
print(bbb.dtype)

bbb= np.array([10,20,3.14])
print(bbb.dtype)
```

```python
# 개수 확인
aa= np.array([10,20,30])
print(len(aa))
print(aa.shape)

bb= np.array([
    [10,20,30,40],
    [5,6,7,8]
])
print(len(bb))
print(bb.shape) #행렬을 튜플로
print(bb.shape[0]) #행개수
print(bb.shape[1]) #열개수

# 전체 요소 개수
print(bb.size)
```

```python
aaa = np.zeros(5)
print(aaa)
print(np.ones(5))
print(np.full(6, 7))
print(np.arange(5))
print(np.arange(2,20,3))
```

```python
#인덱싱과 슬라이싱
scores = np.array([87,87,96,89,78,84,89,95,94,92,])
# 1번학생 점수
first_student = scores[0] # indexing
# 3~7번 학생 점수
middle_students = scores[2:7] # slicing
print(first_student)
print(middle_students)

# 블리언 인덱싱
high_scores = scores[scores>=90] # 조건
print(high_scores)
print(high_scores.mean()) # 평균
```

```python
# 차원 변경
a = np.arange(1,13).reshape(3,4) # reshape(-1,4) 열개수를 기준으로 행을 자동으로 생성
print(a.shape)
print(a)
print(a[0]) # 첫번째 행만 출력
print(a[0:2]) # 첫번째 행부터 2번째 행까지 슬라이싱
print(a[:,2]) # 두번째 열만 모든 행을 포함 하기 위해서 : 만 사용
```

```python
sales_df = pd.DataFrame({
    'date':['2026-01-01']*3 + ['2026-01-02']*3 + ['2026-01-03']*3, # 26.01.01 3개, 02 3개 03 3개 총 9개
    'region':['seoul','busan','daegu'] * 3,
    'salese':[120,90,70,145,105,85,16,115,100],
    'unit':[12,9,7,14,10,8,1,11,10],
    'ad_cost':[20,15,10,8,45,24,np.nan,20,10],
})
sales_df.info()
```

```python
#결측치 확인
sales_df.isna()
```

```python
# 결측치 수량확인
sales_df.isna().sum()
```

```python
#ad_cost(결측치가 있는 )의 평균 구하기
sales_df['ad_cost']=sales_df['ad_cost'].fillna(sales_df['ad_cost'].median())
```

```python
sales_df.isna().sum()
```

```python
# 지역별 평균판매액
region_df = sales_df.groupby('region')['salese'].agg(['sum','mean'])
region_df
```

```python
# 시각화
import matplotlib.pyplot as plt
# 선 그래프
plt.figure(figsize=(7,3))
plt.plot(region_df.index, region_df.values, marker='o')
plt.show()
```

```python
# 산점도 그래프
plt.figure(figsize=(7,3))
plt.scatter(sales_df['ad_cost'], sales_df['salese'], marker='o')
plt.grid(True)
plt.show()
```
